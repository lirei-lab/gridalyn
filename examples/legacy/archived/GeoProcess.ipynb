```python
import geopandas as gpd
from shapely.geometry import Polygon

# Load the Trois-Rivières buildings dataset
buildings_file = "datasets/trois_rivieres_buildings.geojson"
buildings = gpd.read_file(buildings_file)

# Define the selection polygon (your provided data)
polygon_coords = [[-72.62417036110914, 46.34726673598499],
                  [-72.61452837213456, 46.35379678880483],
                  [-72.61013276391213, 46.354794761027705],
                  [-72.58624610343783, 46.339943052357285],
                  [-72.58659312513969, 46.33910452912204],
                  [-72.59730468715948, 46.32824092363754],
                  [-72.6237240054993, 46.34619741748094],
                  [-72.62417036110914, 46.34726673598499]]

# Create a Polygon geometry from the coordinates
selection_polygon = gpd.GeoDataFrame(
    {"geometry": [Polygon(polygon_coords)]},
    crs="EPSG:4326"  # CRS for WGS84
)

# Perform spatial intersection to select buildings inside the polygon
buildings_within_polygon = gpd.overlay(buildings, selection_polygon, how="intersection")

# Save the result to a new GeoJSON file
output_file = "datasets/buildings_inside_polygon.geojson"
buildings_within_polygon.to_file(output_file, driver="GeoJSON")

print(f"Buildings inside the polygon saved to {output_file}")
```
